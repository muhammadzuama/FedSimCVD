```python
import pandas as pd
from pathlib import Path

# Direktori dataset
DATA_DIR = Path(
    r"C:\Users\s2\Documents\FIXRKI\FIXbanget\newscema\Top20\2_per_client\test"
)

# Membaca seluruh file CSV
acc    = pd.read_csv(DATA_DIR / "acc.csv")
auc    = pd.read_csv(DATA_DIR / "auc.csv")
f1     = pd.read_csv(DATA_DIR / "f1.csv")
prauc  = pd.read_csv(DATA_DIR / "prauc.csv")
prec   = pd.read_csv(DATA_DIR / "prec.csv")
rec    = pd.read_csv(DATA_DIR / "rec.csv")

# Dictionary seluruh metrik
metrics = {
    "acc": acc,
    "auc": auc,
    "f1": f1,
    "prauc": prauc,
    "prec": prec,
    "rec": rec
}

# Menampilkan informasi setiap dataset
for name, df in metrics.items():
    print(f"\n{'='*50}")
    print(f"METRIC: {name.upper()}")
    print(f"{'='*50}")
    print(df.shape)
    display(df.head())
```

```python
import pandas as pd
import numpy as np

# Algoritma yang dibandingkan
selected_algorithms = [
    "fedpairwise_T0.01_d0.5",
    "fedpairwise_T0.01_d0.3",
    "fedpairwise_T0.01_d0",
    "fedprox",
    "fedavg"
]

metric_names = ["acc", "auc", "f1", "prauc", "prec", "rec"]

# Menyimpan hasil kemenangan
win_results = {}
winner_details = {}

for metric in metric_names:

    df = filtered_metrics[metric].copy()

    # Menghitung jumlah kemenangan
    wins = pd.Series(0, index=selected_algorithms, dtype=int)
    shared_wins = pd.Series(0.0, index=selected_algorithms)
    ties = 0
    client_winners = []

    for idx, row in df.iterrows():

        row = row.dropna()

        if row.empty:
            continue

        max_value = row.max()

        # Algoritma dengan nilai tertinggi
        winners = row.index[
            np.isclose(row.values, max_value)
        ].tolist()

        client_winners.append({
            "Client": idx + 1,
            "Winner": ", ".join(winners),
            "Number of Winners": len(winners),
            "Best Score": max_value
        })

        # Kemenangan tunggal dan kemenangan bersama
        if len(winners) == 1:
            wins[winners[0]] += 1
        else:
            ties += 1

        for winner in winners:
            shared_wins[winner] += 1 / len(winners)

    win_results[metric] = {
        "Outright Wins": wins,
        "Shared Wins": shared_wins,
        "Ties": ties
    }

    winner_details[metric] = pd.DataFrame(client_winners)

# Menampilkan detail pemenang setiap klien
for metric in metric_names:

    print(f"\n{'='*75}")
    print(f"CLIENT WINNERS - {metric.upper()}")
    print(f"{'='*75}")

    display(winner_details[metric])
```

```python
summary_wins = pd.DataFrame(
    {
        metric: win_results[metric]["Outright Wins"]
        for metric in metric_names
    }
).T

summary_wins.index.name = "Metric"

print("JUMLAH KEMENANGAN SETIAP ALGORITMA")
display(summary_wins)
```

```python
# Total klien valid setiap metrik
total_clients = pd.Series({
    metric: len(winner_details[metric])
    for metric in metric_names
})

win_percentage = summary_wins.div(
    total_clients, axis=0
) * 100

print("PERSENTASE KEMENANGAN (%)")
display(win_percentage.round(2))
```

```python
for metric in metric_names:

    wins = win_results[metric]["Outright Wins"]

    highest = wins.max()
    best_algorithms = wins[wins == highest].index.tolist()

    print(f"\n{metric.upper()}")
    print(f"Jumlah kemenangan tertinggi: {highest}")
    print(f"Algoritma pemenang: {', '.join(best_algorithms)}")
```

```python
overall_wins = summary_wins.sum(axis=0).sort_values(ascending=False)

overall_summary = pd.DataFrame({
    "Total Wins": overall_wins,
    "Win Percentage": (
        overall_wins / summary_wins.values.sum() * 100
    )
})

print("OVERALL CLIENT WINS")
display(overall_summary.round(2))
```
