```python
import pandas as pd
from pathlib import Path

root = Path(r"C:\Users\s2\Documents\FIXRKI\FIXbanget\newscema\Full_Client\5_runs")
frames = []
for d in root.glob("out_*_s*"):
    algo, seed = d.name[4:].rsplit("_s", 1)
    f = d / "eval_test_per_client.csv"
    if f.exists():
        df = pd.read_csv(f)
        df.insert(0, "algoritma", algo)
        df.insert(1, "seed", int(seed))
        frames.append(df)

semua = pd.concat(frames, ignore_index=True)
semua.to_csv(root.parent / "test_per_client_per_seed.csv", index=False)
print(semua.head())
```
