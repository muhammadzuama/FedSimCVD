```python
import pandas as pd
data2022 = pd.read_csv(r"C:\Users\s2\Documents\RKI\BRFSS\brfss2022_diabetes.csv")
data2023 = pd.read_csv(r"C:\Users\s2\Documents\RKI\BRFSS\brfss2023_diabetes.csv")
data2024 = pd.read_csv(r"C:\Users\s2\Documents\RKI\BRFSS\brfss2024_diabetes.csv")
data2025 = pd.read_csv(r"C:\Users\s2\Documents\RKI\BRFSS\brfss2025_diabetes.csv")
```

```python
print("Data 2022:")
data2022.info()
print("Data 2023:")
data2023.info()
print("Data 2024:")
data2024.info()
print("Data 2025:")
data2025.info()
```

```python
data2022 = data2022.drop(columns=["PRIMINSR", "SLEPTIM1", "_DRNKWK2"])
data2023 = data2023.drop(columns=["_DRNKWK2"])
# data2025 = data2025.drop(columns=["SLEPTIM1"])
# data2024 sudah pas
```

```python
data_combined = pd.concat([data2022, data2023, data2024, data2025], ignore_index=True)
```

```python
data_combined.to_csv(r"C:\Users\s2\Documents\RKI\BRFSS\brfss2022-2025_diabetes.csv", index=False)
```

```python
# %% Loop semua state
for state, g in data_combined.groupby("_STATE"):
    print(f"\n=== State {int(state)} | total: {len(g)} ===")
    print(g["IYEAR"].value_counts().sort_index().to_string())
```

```python
# %% Setup
import matplotlib.pyplot as plt
import seaborn as sns

# Gabungkan 2026 ke 2025 supaya tidak jadi grup sendiri
df = data_combined.copy()
df["IYEAR"] = df["IYEAR"].replace(2026, 2025)
df["_STATE"] = df["_STATE"].astype(int)

ct = pd.crosstab(df["_STATE"], df["IYEAR"])
```

```python
# %% Heatmap state x tahun
plt.figure(figsize=(7, 14))
sns.heatmap(ct, annot=True, fmt="d", cmap="YlGnBu",
            cbar_kws={"label": "Jumlah responden"}, linewidths=.3)
plt.title("Jumlah responden diabetes per state per tahun")
plt.xlabel("Tahun")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```

```python
# %% Line plot per state (semua state, garis tipis)
plt.figure(figsize=(10, 6))
for s in ct.index:
    plt.plot(ct.columns, ct.loc[s], marker="o", alpha=.5, linewidth=1)
plt.axhline(50, color="red", linestyle="--", label="Batas 50 sampel")
plt.xticks(ct.columns)
plt.title("Jumlah sampel tiap state per tahun")
plt.xlabel("Tahun"); plt.ylabel("Jumlah responden")
plt.legend()
plt.show()
```
